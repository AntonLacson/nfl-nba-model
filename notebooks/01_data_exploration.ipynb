## Import

```python
import pandas as pd 
import numpy as np 
```

## Load in NFL Data

```python
url = "https://raw.githubusercontent.com/nflverse/nfldata/master/data/games.csv"
games = pd.read_csv(url)
```

```python
games.head()
```

```python
games.shape
```

```python
games.columns.tolist()
```

## Set to the important columns 

```python
games [[
    "season", 
    "week", 
    "gameday",
    "away_team",
    "away_score",
    "home_team",
    "home_score",
    "spread_line",
    "total_line"
]].head(10)
```

```python
## Filter to 2018 - 2026
nfl = games[
    (games["season"] >= 2015) & 
    (games["season"] <= 2025)
].copy()
```

```python
nfl.shape
```

```python
nfl["margin"] = nfl["home_score"] - nfl["away_score"]
```

```python
nfl[[
    "season",
    "week",
    "away_team",
    "away_score",
    "home_team",
    "home_score",
    "margin",
    "spread_line"
]].head(10)
```
